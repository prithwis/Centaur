## Event

```python
E = []
```

```python
event = Event(
    actor="Iran",
    action=(
        "Iran declares the Strait of Hormuz closed to all commercial shipping, "
        "orders its forces to prevent tanker transit, and launches attacks against "
        "US naval forces operating in the Gulf."
    )
)

event.event_id = f"E{len(E)}"
E.append(event)
E[0].show()
```

```python
ZW_Update_Prompt = """
You are updating a simulated WorldState after an executed Event.

The CURRENT WORLDSTATE is authoritative. It contains the complete state
of the simulated world immediately before the Event.

The EXECUTED EVENT has occurred. Do not assess whether the Event was
likely, wise, desirable, or strategically sensible. Your task is only
to determine its consequences.

STATE TRANSITION RULES

1. Begin strictly from the CURRENT WORLDSTATE. Do not introduce external
   real-world information or developments not contained in the current
   state or reasonably implied by the Event.

2. Examine every WorldVariable for possible causal impact from the Event.

3. Change a WorldVariable only when there is a reasonable causal pathway
   from the Event to that variable.

4. Causal effects may be direct or indirect. Follow important secondary
   consequences where they reasonably follow from the Event and the
   current state.

5. Do not create unrelated consequences merely to make the simulation
   more eventful.

6. For an affected WorldVariable:
   - adjust its index in the direction and magnitude justified by the Event;
   - rewrite status to describe its new state;
   - write rationale explaining why the index changed from its previous
     value to the new value;
   - update conditions to preserve currently operative factors.

7. For an unaffected WorldVariable:
   - preserve its index;
   - preserve its status;
   - preserve its conditions;
   - write rationale briefly explaining why the Event does not justify
     changing the index.

8. Existing conditions are state memory. Preserve them unless the Event
   changes, resolves, supersedes, or makes them irrelevant.

9. Preserve every WorldVariable's:
   - name
   - description
   - direction

10. The index value 100 represents the W0 baseline for that variable.
    It does NOT represent normality, neutrality, equilibrium, or a
    historically average condition.

    Subsequent index values represent movement relative to that W0 baseline.

    Changes should be proportionate to the significance of the Event.
    Avoid exaggerated movements unless the Event represents a genuinely
    major structural change.

RATIONALE

11. Rationale must explain the numerical transition of each WorldVariable.

    For an affected variable, explicitly consider:
    - its previous index;
    - the causal effect of the Event;
    - relevant operative conditions in the CURRENT WORLDSTATE;
    - why the resulting direction and approximate magnitude of change
      are justified.

    Rationale should therefore make clear why a transition such as
    55 -> 62, 100 -> 115, or 120 -> 112 was chosen.

    Do not claim that the index is an empirically measured quantity or
    that its exact numerical value has statistical precision.

12. Rationale must not introduce new facts. Every factual premise used
    in the rationale must come from the CURRENT WORLDSTATE, the EXECUTED
    EVENT, or a reasonable causal inference from those inputs.

13. Keep rationale concise. It is an audit explanation for the index
    transition, not a general discussion or prediction.

OUTPUT

14. Return the COMPLETE next WorldState containing every WorldVariable,
    including variables that did not change.

15. Return only valid JSON.
    Do not use markdown code fences.
    Do not include commentary before or after the JSON.
"""
```

```python
_ = ZeitWorld.update( W, E[0], ZW_Update_Prompt)
```
